## Regression Model Evaluation

```python
column_names = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT', 'MEDV']
data = pd.read_csv('housing.csv', header=None, delimiter=r"\s+", names=column_names)
```

```python
"""
CRIM - per capita crime rate by town
ZN - proportion of residential land zoned for lots over 25,000 sq.ft.
INDUS - proportion of non-retail business acres per town.
CHAS - Charles River dummy variable (1 if tract bounds river; 0 otherwise)
NOX - nitric oxides concentration (parts per 10 million)
RM - average number of rooms per dwelling
AGE - proportion of owner-occupied units built prior to 1940
DIS - weighted distances to five Boston employment centres
RAD - index of accessibility to radial highways
TAX - full-value property-tax rate per $10,000
PTRATIO - pupil-teacher ratio by town
B - 1000(Bk - 0.63)^2 where Bk is the proportion of blacks by town
LSTAT - % lower status of the population
MEDV - Median value of owner-occupied homes in $1000's"""
```

```python
data
```

## 1. Split this data set into training (80%) and testing (20%) sets.

The `MEDV` field represents the median value of owner-occupied homes (in $1000's) and is the target variable that we will want to predict.

```python
from sklearn.model_selection import train_test_split

# Features (all columns except MEDV)
X_data = data.drop('MEDV', axis=1)

# Target
y_data = data['MEDV']

# Train-test split (80% train, 20% test)
X_train_data, X_test_data, y_train_data, y_test_data = train_test_split(X_data, y_data, test_size=0.2, random_state=42)

print("X_train_data.shape\n", X_train_data.shape, "\n\nX_test_data.shape\n", X_test_data.shape, "\n\ny_train_data.shape\n", y_train_data.shape, "\n\ny_test_daya.shape\n", y_test_data.shape)
```

## 2. Train a `LinearRegression` model on this data set and generate predictions on both the training and the testing set.

```python
from sklearn.linear_model import LinearRegression

# Create the model
model_data = LinearRegression()

# Train the model
model_data.fit(X_train_data, y_train_data)

# Predictions
y_train_pred_data = model_data.predict(X_train_data)
y_test_pred_data = model_data.predict(X_test_data)

print("y_train_pred_data[:5]\n", y_train_pred_data[:5], "\n\ny_test_pred_data[:5]\n", y_test_pred_data[:5])
```

## 3. Calculate and print R-squared for both the training and the testing set.

```python
from sklearn.metrics import r2_score

# R-squared scores
r2_train_data = r2_score(y_train_data, y_train_pred_data)
r2_test_data = r2_score(y_test_data, y_test_pred_data)

print("R-squared (train):", r2_train_data, "\n")
print("R-squared (test):", r2_test_data, "\n")
```

## 4. Calculate and print mean squared error for both the training and the testing set.

```python
from sklearn.metrics import mean_squared_error

mse_train_data = mean_squared_error(y_train_data, y_train_pred_data)
mse_test_data = mean_squared_error(y_test_data, y_test_pred_data)

print("MSE (train):", mse_train_data, "\n")
print("MSE (test):", mse_test_data, "\n")
```

## 5. Calculate and print mean absolute error for both the training and the testing set.

```python
from sklearn.metrics import mean_absolute_error

mae_train_data = mean_absolute_error(y_train_data, y_train_pred_data)
mae_test_data = mean_absolute_error(y_test_data, y_test_pred_data)

print("MAE (train):", mae_train_data, "\n")
print("MAE (test):", mae_test_data, "\n")
```

## Classification Model Evaluation

```python
from sklearn.datasets import load_iris
data = load_iris()
```

## 6. Split this data set into training (80%) and testing (20%) sets.

The `class` field represents the type of flower and is the target variable that we will want to predict.

```python
from sklearn.model_selection import train_test_split

X_iris = data.data          # features
y_iris = target        # class labels

X_train_iris, X_test_iris, y_train_iris, y_test_iris = train_test_split(
    X_iris, y_iris, test_size=0.2, random_state=42
)

print("X_train_iris.shape\n", X_train_iris.shape, "\n\nX_test_iris.shape\n", X_test_iris.shape)
```

## 7. Train a `LogisticRegression` model on this data set and generate predictions on both the training and the testing set.

```python
from sklearn.linear_model import LogisticRegression

# Create the model
model_iris = LogisticRegression(max_iter=200) 
# The default (100) triggers a convergence warning on the Iris dataset.
# Setting it to 200 ensures the model converges cleanly.

# Train the model
model_iris.fit(X_train_iris, y_train_iris)

# Predictions
y_train_pred_iris = model_iris.predict(X_train_iris)
y_test_pred_iris = model_iris.predict(X_test_iris)

print("y_train_pred_iris[:5] : ", y_train_pred_iris[:5], "\n\ny_test_pred_iris[:5] : ", y_test_pred_iris[:5])
```

## 8. Calculate and print the accuracy score for both the training and the testing set.

```python
from sklearn.metrics import accuracy_score

accuracy_train_iris = accuracy_score(y_train_iris, y_train_pred_iris)
accuracy_test_iris = accuracy_score(y_test_iris, y_test_pred_iris)

print("Accuracy (train):", accuracy_train_iris, "\n")
print("Accuracy (test):", accuracy_test_iris, "\n")
```

## 9. Calculate and print the balanced accuracy score for both the training and the testing set.

```python
from sklearn.metrics import balanced_accuracy_score

balanced_acc_train_iris = balanced_accuracy_score(y_train_iris, y_train_pred_iris)
balanced_acc_test_iris = balanced_accuracy_score(y_test_iris, y_test_pred_iris)

print("Balanced Accuracy (train):", balanced_acc_train_iris, "\n")
print("Balanced Accuracy (test):", balanced_acc_test_iris, "\n")
```

## 10. Calculate and print the precision score for both the training and the testing set.

```python
from sklearn.metrics import precision_score

precision_train_iris = precision_score(y_train_iris, y_train_pred_iris, average='macro')
precision_test_iris = precision_score(y_test_iris, y_test_pred_iris, average='macro')

print("Precision (train):", precision_train_iris, "\n")
print("Precision (test):", precision_test_iris, "\n")
```

## 11. Calculate and print the recall score for both the training and the testing set.

```python
from sklearn.metrics import recall_score

recall_train_iris = recall_score(y_train_iris, y_train_pred_iris, average='macro')
recall_test_iris = recall_score(y_test_iris, y_test_pred_iris, average='macro')

print("Recall (train):", recall_train_iris, "\n")
print("Recall (test):", recall_test_iris, "\n")
```

## 12. Calculate and print the F1 score for both the training and the testing set.

```python
from sklearn.metrics import f1_score

f1_train_iris = f1_score(y_train_iris, y_train_pred_iris, average='macro')
f1_test_iris = f1_score(y_test_iris, y_test_pred_iris, average='macro')

print("F1-score (train):", f1_train_iris, "\n")
print("F1-score (test):", f1_test_iris, "\n")
```

## 13. Generate confusion matrices for both the training and the testing set.

```python
from sklearn.metrics import confusion_matrix

cm_train_iris = confusion_matrix(y_train_iris, y_train_pred_iris)
cm_test_iris = confusion_matrix(y_test_iris, y_test_pred_iris)

print("Confusion Matrix (train):\n", cm_train_iris, "\n")
print("\nConfusion Matrix (test):\n", cm_test_iris, "\n")
```

### **Regression Models (Boston Housing)**
#### **Models worth testing**
* **Linear Regression**
* **Ridge Regression**
* **Lasso Regression**
* **Decision Tree Regressor**
* **Random Forest Regressor**
* **KNN Regressor**

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

models_boston_housing = {
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(),
    "Lasso": Lasso(),
    "Decision Tree": DecisionTreeRegressor(),
    "Random Forest": RandomForestRegressor(),
    "KNN": KNeighborsRegressor()
}

results_boston_housing = []

for name, model in models_boston_housing.items():
    model.fit(X_train_data, y_train_data)
    y_pred_boston_housing = model.predict(X_test_data)

    mse_boston_housing = mean_squared_error(y_test_data, y_pred_boston_housing)
    rmse_boston_housing = np.sqrt(mse_boston_housing)
    mae_boston_housing = mean_absolute_error(y_test_data, y_pred_boston_housing)
    r2_boston_housing = r2_score(y_test_data, y_pred_boston_housing)

    results_boston_housing.append([name, mse_boston_housing, rmse_boston_housing, mae_boston_housing, r2_boston_housing])

import pandas as pd
pd.DataFrame(results_boston_housing, columns=["Model", "MSE", "RMSE", "MAE", "R2"])
```

### **Classification Models (Iris)**
#### **Models worth testing**
* **Logistic Regression**
* **KNN Classifier**
* **Decision Tree Classifier**
* **Random Forest Classifier**
* **SVM Classifier**

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

models_iris = {
    "Logistic Regression": LogisticRegression(max_iter=200),
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "SVM": SVC()
}

results_iris = []

for name, model in models_iris.items():
    model.fit(X_train_iris, y_train_iris)
    y_pred_iris = model.predict(X_test_iris)

    acc_iris = accuracy_score(y_test_iris, y_pred_iris)
    precision_iris = precision_score(y_test_iris, y_pred_iris, average='macro')
    recall_iris = recall_score(y_test_iris, y_pred_iris, average='macro')
    f1_iris = f1_score(y_test_iris, y_pred_iris, average='macro')

    results_iris.append([name, acc_iris, precision_iris, recall_iris, f1_iris])

import pandas as pd
pd.DataFrame(results_iris, columns=["Model", "Accuracy", "Precision", "Recall", "F1"])
```
